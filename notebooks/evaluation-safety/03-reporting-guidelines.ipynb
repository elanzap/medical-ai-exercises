**In short:** Reporting guidelines are checklists that say what a study must report so readers can judge it. For clinical AI there is now one for almost every study type: TRIPOD+AI for prediction models, STARD-AI for diagnostic accuracy, CLAIM for imaging AI, DECIDE-AI for early clinical evaluation, SPIRIT-AI and CONSORT-AI for trials, TRIPOD-LLM for studies of language models and CHART for chatbot health-advice studies. PROBAST+AI helps readers judge risk of bias. Using them makes studies more transparent and missing information easier to spot.
