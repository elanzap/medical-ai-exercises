### Loading and summarising a table

Load six synthetic patients, describe the numeric columns and count missing values.

````python
from io import StringIO
import pandas as pd

csv_text = """patient_id,age,sex,hba1c,systolic_bp
s01,54,F,7.9,138
s02,61,M,6.4,142
s03,47,F,8.6,129
s04,70,M,,151
s05,58,F,6.9,134
s06,66,M,9.1,147
"""
df = pd.read_csv(StringIO(csv_text))   # with a real file: pd.read_csv("labs.csv")

print(df.shape)
print(df.describe().round(1))
print("\nMissing values:\n", df.isna().sum())

# (6, 5)
#         age  hba1c  systolic_bp
# count   6.0    5.0          6.0
# mean   59.3    7.8        140.2
# std     8.3    1.1          8.2
# min    47.0    6.4        129.0
# 25%    55.0    6.9        135.0
# 50%    59.5    7.9        140.0
# 75%    64.8    8.6        145.8
# max    70.0    9.1        151.0
#
# Missing values:
#  patient_id     0
# age            0
# sex            0
# hba1c          1
# systolic_bp    0
````

### Grouping, new columns and a hidden trap

Continuing from the previous block.

````python
df["hba1c_above_7"] = df.hba1c > 7.0
print(df.groupby("sex")[["hba1c", "systolic_bp"]].mean().round(1))
print(df[["patient_id", "hba1c", "hba1c_above_7"]])

#      hba1c  systolic_bp
# sex
# F      7.8        133.7
# M      7.8        146.7
#
#   patient_id  hba1c  hba1c_above_7
# 0        s01    7.9           True
# 1        s02    6.4          False
# 2        s03    8.6           True
# 3        s04    NaN          False
# 4        s05    6.9          False
# 5        s06    9.1           True
````

### A first chart

Continuing from the previous block: a scatter plot with a reference line, saved as an image.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 3.5))
ax.scatter(df.age, df.hba1c)
ax.axhline(7.0, linestyle="--", linewidth=1)
ax.set_xlabel("Age (years)")
ax.set_ylabel("HbA1c (%)")
ax.set_title("Synthetic patients: HbA1c by age")
fig.tight_layout()
fig.savefig("hba1c_by_age.png", dpi=150)
````
