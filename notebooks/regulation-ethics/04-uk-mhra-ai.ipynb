**In short:** In Great Britain, AI medical devices are regulated by the MHRA under the UK Medical Devices Regulations 2002, which were not written with AI in mind. The MHRA has been reforming software regulation, testing approaches in its AI Airlock sandbox, and in 2026 the government accepted all 44 recommendations of the National Commission into the Regulation of AI in Healthcare. Draft guidance on change control is due by December 2026 and a consultation on how to qualify and classify AI devices in 2027. NHS organisations also apply their own clinical safety and evidence standards before adopting tools.
