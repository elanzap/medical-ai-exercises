**In short:** When AI contributes to harm, responsibility can fall on the clinician (judged against the standard of care), the organisation that deployed the tool, and the manufacturer (through product liability). In the EU, the revised Product Liability Directive explicitly covers software, including AI, for products placed on the market from 9 December 2026. Patients should generally be told when AI plays a meaningful role in their care, and in some places that is a legal requirement for AI-generated communications and chatbots. Clear documentation, contracts and governance protect both patients and clinicians.
