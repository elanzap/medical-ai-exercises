**In short:** In India, software with a medical purpose is a medical device regulated by the Central Drugs Standard Control Organisation (CDSCO) under the Medical Devices Rules, 2017, which sort devices into risk classes A (low) to D (high). Alongside device rules, the ICMR's 2023 ethical guidelines for AI in healthcare, the DPDP Act for personal data, and telemedicine and professional-conduct rules shape how AI can be used in Indian care.
