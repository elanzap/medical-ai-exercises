**In short:** Software as a Medical Device (SaMD) is software intended for a medical purpose, such as diagnosing, monitoring or guiding treatment, that achieves that purpose without being part of a hardware device. Whether software is a medical device depends on its intended purpose, as stated by the manufacturer, not on whether it uses AI. Risk depends on how serious the condition is and how much the software's output drives clinical decisions, and higher risk means more evidence and oversight.

### Categorising risk

The IMDRF SaMD risk categories, applied to four illustrative diabetes-related tools.

```python
# IMDRF SaMD risk categories (N12, 2014): state of the healthcare situation x significance of the information
CATEGORY = {
    ("critical", "treat or diagnose"): "IV", ("critical", "drive clinical management"): "III",
    ("critical", "inform clinical management"): "II",
    ("serious", "treat or diagnose"): "III", ("serious", "drive clinical management"): "II",
    ("serious", "inform clinical management"): "I",
    ("non-serious", "treat or diagnose"): "II", ("non-serious", "drive clinical management"): "I",
    ("non-serious", "inform clinical management"): "I",
}

examples = [
    ("Autonomous diabetic retinopathy screening result", "serious", "treat or diagnose"),
    ("Insulin dose calculator for type 1 diabetes", "critical", "drive clinical management"),
    ("Sepsis early-warning score shown to ward staff", "critical", "inform clinical management"),
    ("App summarising home glucose readings for the clinician", "non-serious", "inform clinical management"),
]
for name, situation, significance in examples:
    print(f"Category {CATEGORY[(situation, significance)]:<3} {name}  ({situation}; {significance})")

# Category III Autonomous diabetic retinopathy screening result  (serious; treat or diagnose)
# Category III Insulin dose calculator for type 1 diabetes  (critical; drive clinical management)
# Category II  Sepsis early-warning score shown to ward staff  (critical; inform clinical management)
# Category I   App summarising home glucose readings for the clinician  (non-serious; inform clinical management)
```
